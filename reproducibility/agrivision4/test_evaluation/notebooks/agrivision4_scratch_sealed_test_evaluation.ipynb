# AgriVision4 Scratch Models: Research V2 Final Holdout

This notebook evaluates the five frozen three-seed scratch comparisons. Run `preflight` first. The sealed mode is one-time, performs no model selection, and refuses to overwrite completed results.

```python
# Configuration: RUN_MODE supports preflight, sealed_evaluation, report_existing.
RUN_MODE = "sealed_evaluation"
ACKNOWLEDGEMENT = "RUN_AGRIVISION4_RESEARCH_V2_FINAL_HOLDOUT_ONCE"
SEALED_ACKNOWLEDGEMENT = "RUN_AGRIVISION4_RESEARCH_V2_FINAL_HOLDOUT_ONCE"

DEVICE = "auto"
BATCH_SIZE = 32
NUM_WORKERS = 0
VERIFY_ARCHIVE_HASH = True
VERIFY_IMAGE_HASHES = True
OUTPUT_DIRECTORY = "agrivision4/results/sealed_test/agrivision4_scratch_research_v2_final_holdout_v1"
```

```python
from pathlib import Path
import sys

def find_repo_root(start: Path) -> Path:
    candidates = [start.resolve(), *start.resolve().parents]
    for candidate in candidates:
        marker = candidate / "code" / "agrivision4" / "evaluation" / "sealed_test.py"
        if marker.is_file():
            return candidate
    raise FileNotFoundError("Run this notebook inside the Agrivision4 workspace")

REPO_ROOT = find_repo_root(Path.cwd())
CODE_ROOT = REPO_ROOT / "code"
if str(CODE_ROOT) not in sys.path:
    sys.path.insert(0, str(CODE_ROOT))
print({"repo_root": str(REPO_ROOT), "run_mode": RUN_MODE, "device": DEVICE})
```

## Execute

In sealed mode, this cell validates the complete frozen contract before opening the final-holdout images. Progress is printed after every checkpoint.

```python
from agrivision4.evaluation.sealed_test import execute_from_values

result = execute_from_values({
    "REPO_ROOT": REPO_ROOT,
    "RUN_MODE": RUN_MODE,
    "ACKNOWLEDGEMENT": ACKNOWLEDGEMENT,
    "DEVICE": DEVICE,
    "BATCH_SIZE": BATCH_SIZE,
    "NUM_WORKERS": NUM_WORKERS,
    "VERIFY_ARCHIVE_HASH": VERIFY_ARCHIVE_HASH,
    "VERIFY_IMAGE_HASHES": VERIFY_IMAGE_HASHES,
    "OUTPUT_DIRECTORY": OUTPUT_DIRECTORY,
})
print({
    "status": result["status"],
    "sealed_inference": result.get("sealed_inference_executed", False),
})
```

## Principal Results

These tables remain in the executed notebook. Machine-readable versions are saved in the result directory.

```python
import pandas as pd
from IPython.display import Markdown, display

if RUN_MODE == "preflight":
    protocol = result["protocol"]
    display(Markdown("### Preflight passed: no sealed inference executed"))
    display(pd.DataFrame([{
        "final_holdout_samples": protocol["final_holdout_samples"],
        "classes": protocol["class_count"],
        "minimum_class_support": protocol["minimum_class_support"],
        "checkpoint_count": result["checkpoint_count"],
        "strict_loads": len(result["strict_load_audits"]),
    }]))
    checkpoint_table = pd.DataFrame(result["checkpoint_table"])
    display(checkpoint_table[[
        "display_name", "model_key", "seed", "parameters",
        "actual_sha256", "test_accessed"
    ]])
else:
    display(Markdown("### Final aggregate ranking: mean, SD, and 95% t CI"))
    aggregate_table = result["aggregate"]
    display(aggregate_table)
    display(Markdown("### Per-seed sealed-test metrics"))
    seed_metrics_table = result["seed_metrics"]
    display(seed_metrics_table)
```
